```kotlin
import java.text.DecimalFormat

fun formatToEngineering(number: Double, precision: Int): String {
    // Determine the exponent based on multiples of 3
    var exponent = 0
    var tempNumber = number
    while (Math.abs(tempNumber) >= 1000 && tempNumber != 0.0) {
        tempNumber /= 1000
        exponent += 3
    }
    while (Math.abs(tempNumber) < 1 && tempNumber != 0.0) {
        tempNumber *= 1000
        exponent -= 3
    }

    // Format the mantissa
    val pattern = "0." + "#".repeat(precision) // e.g., "0.###" for precision 3
    val decimalFormat = DecimalFormat(pattern)
    val formattedMantissa = decimalFormat.format(tempNumber)

    // Construct the final string
    return if (exponent == 0) {
        formattedMantissa
    } else {
        "$formattedMantissa E$exponent"
    }
}

val goalHeight = 98.75 / 100.0 // metres
val ballWeight = 75.0 / 1000.0 // kilograms
val gravity = -9.8 // m/s2

/**
 * Velocity needed to reach *height* meters if thrown at *angle* radians
 * y = y0 + v0t + g*t^2
 */

// Dimensions check:  [a] = [v] / [t] => [v] = [s] / [t] => [a] = [s] / [t]^2
fun velocity(height: Double, angle: Double, gravity: Double): Double {
    return sqrt(2 * -gravity * height / sin(angle) * sin(angle))
}

val minimumVelocity = velocity(goalHeight, 45.0 * PI / 180.0, gravity)

println("v min: ${round(minimumVelocity * 100.0) / 100.0} m/s")

val minimumEnergy = 1.0 / 2.0 * ballWeight * minimumVelocity * minimumVelocity
val decimal = DecimalFormat("#,###.##")
println(
    "Energy: ${
        decimal.format(minimumEnergy)
    } J"
)
println("        ${formatToEngineering(minimumEnergy / 3.6e3, 2)} Wh")

val omega = 6000.0 / 60.0 // revolutions per second
println("Wheel Omega: $omega rps")
//val radialVelocity = omega * radius

val radialVelocity = omega * 2 * PI
println("Radial velocity: $radialVelocity rad/s")

val radius = minimumVelocity / radialVelocity
println("Radius: ${round(radius * 1000.0)} mm")
```
